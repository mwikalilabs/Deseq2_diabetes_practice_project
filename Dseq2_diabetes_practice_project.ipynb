```python
dds = DeseqDataSet(counts=counts_for_deseq, metadata=metadata, design="~group")
```

```python
#size factors
dds.obs["size_factors"]
```

#### does expression differ between t2d and normal

```python
#DeseqStats -- used to compute differential expression statistics
from pydeseq2.ds import DeseqStats
```

```python
stat_results = DeseqStats(dds,contrast=["group","T2D","normal"])
```

```python
#statistical tests results summary 
stat_results.summary()
```

```python
# Extract the full results table as a Pandas DataFrame
results = stat_results.results_df
results.head()
```

```python
(results["padj"] < 0.05).sum()
```

```python
sig = results[results["padj"] < 0.05].sort_values("padj")
sig.head(10)
```

```python
(sig["log2FoldChange"] > 0).sum(), (sig["log2FoldChange"] < 0).sum()
```
